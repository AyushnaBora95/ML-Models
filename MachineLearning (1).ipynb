```python
# These are the headers which you need to link to your dataset.
headers = ["symboling","normalized-losses","make","fuel-type","aspiration", "num-of-doors","body-style",
         "drive-wheels","engine-location","wheel-base", "length","width","height","curb-weight","engine-type",
         "num-of-cylinders", "engine-size","fuel-system","bore","stroke","compression-ratio","horsepower",
         "peak-rpm","city-mpg","highway-mpg","price"]
```

```python
df=pd.read_csv("car_dataset.data", names=headers)
print(df)
```

```python
df.replace("?", np.nan, inplace=True)
df.replace("none", np.nan, inplace=True)
print(df)
df.head(5)
```

```python
missing_data=df.isnull()
print(missing_data)
```

```python
for column in headers:
  print(column)
  print(missing_data[column].value_counts())
  print("")
```

```python
avg=df["normalized-losses"].astype(float).mean(axis=0)
print("average normalized-losses are", avg)
```

```python
df["normalized-losses"].replace(np.nan, avg, inplace = True)
```

```python
avg_bore=df["bore"].astype(float).mean(axis=0)
print("average bore are", avg_bore)
```

```python
df["bore"].replace(np.nan, avg, inplace=True)
```

```python
avg_stroke=df["stroke"].astype(float).mean(axis=0)
print("average stroke are: ", avg_stroke)
```

```python
df.replace(np.nan, avg_stroke, inplace=True)
```

```python
avg_horsepower=df["horsepower"].astype("float").mean(axis=0)
print("average horsepower are: ", avg_horsepower)
```

```python
df['horsepower'].replace(np.nan,avg_horsepower, inplace=True)
```
